```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("MongoSparkTest") \
    .master("local[*]") \
    .config("spark.jars.packages",
            "org.mongodb.spark:mongo-spark-connector_2.12-10.3.0,"
            "org.mongodb:mongodb-driver-sync:4.11.5,"
            "org.mongodb:mongodb-driver-core:4.11.5,"
            "org.mongodb:bson:4.11.5") \
    .config("spark.mongodb.read.connection.uri",
            "mongodb://admin:admin123@localhost:27017/testdb?authSource=admin") \
    .config("spark.mongodb.write.connection.uri",
            "mongodb://admin:admin123@localhost:27017/testdb?authSource=admin") \
    .getOrCreate()

spark.sparkContext.setLogLevel("OFF")

# Testar leitura
df = spark.read.format("mongodb") \
    .option("database", "testdb") \
    .option("collection", "clientes") \
    .load()

df.show()
```
